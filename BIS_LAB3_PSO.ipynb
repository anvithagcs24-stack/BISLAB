```python
import random

# Number of applications
n = 4
# Total available mobile data
total_data = 20
# Minimum data required by each application
min_data = [2, 3, 1, 2]
# Maximum data allowed for each application
max_data = [6, 7, 5, 6]
# Priority of each application
priority = [3, 2, 1, 2]

# PSO parameters
particles = 10
iterations = 50
w = 0.5
c1 = 1.5
c2 = 1.5

# Create particles randomly
position = []
velocity = []
for i in range(particles):
    p = []
    v = []
    for j in range(n):
        p.append(random.uniform(min_data[j], max_data[j]))
        v.append(0)
    position.append(p)
    velocity.append(v)

# Fitness function
def fitness(p):
    total = sum(p)
    cost = 0
    # Penalty if total data exceeds available data
    if total > total_data:
        cost += (total - total_data) * 100
    # Penalty if application gets less than minimum
    for i in range(n):
        if p[i] < min_data[i]:
            cost += (min_data[i] - p[i]) * 100 * priority[i]
        # Penalty for unnecessary excess data
        if p[i] > min_data[i]:
            cost += p[i] - min_data[i]
    return cost

# Initially set personal best
pbest = []
for i in range(particles):
    pbest.append(position[i].copy())

# Find initial global best
gbest = pbest[0].copy()
for i in range(particles):
    if fitness(pbest[i]) < fitness(gbest):
        gbest = pbest[i].copy()

# PSO main loop
for iteration in range(iterations):
    for i in range(particles):
        for j in range(n):
            r1 = random.random()
            r2 = random.random()
            # Update velocity
            velocity[i][j] = (
                w * velocity[i][j]
                + c1 * r1 * (pbest[i][j] - position[i][j])
                + c2 * r2 * (gbest[j] - position[i][j])
            )
            # Update position
            position[i][j] += velocity[i][j]
            # Keep position within minimum and maximum limits
            if position[i][j] < min_data[j]:
                position[i][j] = min_data[j]
            if position[i][j] > max_data[j]:
                position[i][j] = max_data[j]
        # Update personal best
        if fitness(position[i]) < fitness(pbest[i]):
            pbest[i] = position[i].copy()
        # Update global best
        if fitness(pbest[i]) < fitness(gbest):
            gbest = pbest[i].copy()

print("Optimal Data Allocation:")
for i in range(n):
    print("Application", i + 1, ":", round(gbest[i], 2), "GB")
print("Total Data Used:", round(sum(gbest), 2), "GB")
print("Fitness:", round(fitness(gbest), 2))
```
