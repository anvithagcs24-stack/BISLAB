```python
import random

# Subjects
subjects = ["S1", "S2", "S3", "S4", "S5"]

# Conflicting subject pairs
conflicts = [
    ("S1", "S2"),
    ("S1", "S3"),
    ("S2", "S4"),
    ("S3", "S5"),
    ("S4", "S5")
]

# Time slots and halls
slots = ["T1", "T2", "T3"]
halls = ["H1", "H2"]

# Genetic Algorithm parameters
population_size = 10
mutation_rate = 0.1
crossover_rate = 0.8
generations = 50


# Create a random timetable for one individual
def create_individual():
    individual = []

    for subject in subjects:
        slot = random.choice(slots)
        hall = random.choice(halls)
        individual.append((slot, hall))

    return individual


# Create initial population
population = []

for i in range(population_size):
    population.append(create_individual())


# Fitness function
def fitness(individual):

    penalty = 0

    # Check conflicting subjects
    for s1, s2 in conflicts:

        i = subjects.index(s1)
        j = subjects.index(s2)

        # If conflicting subjects have same time slot
        if individual[i][0] == individual[j][0]:
            penalty += 1

    # Check if two exams are in same hall at same time
    for i in range(len(subjects)):
        for j in range(i + 1, len(subjects)):

            if individual[i] == individual[j]:
                penalty += 1

    # Higher fitness = better solution
    return 1 / (1 + penalty)


# Roulette-wheel selection
def select_parent(population):

    fitness_values = []

    for individual in population:
        fitness_values.append(fitness(individual))

    return random.choices(population, weights=fitness_values, k=1)[0]


# Single-point crossover
def crossover(parent1, parent2):

    point = random.randint(1, len(subjects) - 1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2


# Mutation
def mutation(individual):

    for i in range(len(individual)):

        if random.random() < mutation_rate:

            slot = random.choice(slots)
            hall = random.choice(halls)

            individual[i] = (slot, hall)

    return individual


# Genetic Algorithm
best_individual = None
best_fitness = 0

for generation in range(generations):

    # Evaluate fitness
    for individual in population:

        current_fitness = fitness(individual)

        if current_fitness > best_fitness:
            best_fitness = current_fitness
            best_individual = individual.copy()

    # Create new population
    new_population = []

    while len(new_population) < population_size:

        # Select parents
        parent1 = select_parent(population)
        parent2 = select_parent(population)

        # Crossover
        if random.random() < crossover_rate:

            child1, child2 = crossover(parent1, parent2)

        else:

            child1 = parent1.copy()
            child2 = parent2.copy()

        # Mutation
        child1 = mutation(child1)
        child2 = mutation(child2)

        # Add children
        new_population.append(child1)

        if len(new_population) < population_size:
            new_population.append(child2)

    # Replace old population
    population = new_population


# Display final answer
print("Best Timetable:")

for i in range(len(subjects)):
    print(subjects[i], "->", best_individual[i])

print("Final Fitness:", best_fitness)
```
